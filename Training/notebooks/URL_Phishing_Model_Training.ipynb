```python
import re
import ipaddress

from urllib.parse import urlsplit

URL_FEATURE_COLUMNS = [
    "url_length",
    "domain_length",
    "path_length",
    "query_length",
    "num_subdomains",
    "has_ip_address",
    "has_https",
    "has_suspicious_tld",
    "num_special_chars",
    "num_digits_in_domain",
    "has_brand_keyword",
    "brand_in_subdomain",
    "has_redirect",
    "path_depth"
]

SUSPICIOUS_TLDS = {
    "tk", "ml", "ga", "cf", "gq",
    "xyz", "top", "click", "pw"
}

BRAND_KEYWORDS = [
    "paypal", "google", "amazon", "apple",
    "microsoft", "facebook", "instagram",
    "whatsapp", "netflix", "sbi",
    "hdfc", "icici", "axisbank",
    "paytm", "phonepe", "visa"
]

print("URL Feature Configuration Ready!")
print("Total Features:", len(URL_FEATURE_COLUMNS))
```

```python
def extract_url_features(url):

    url = str(url).strip()
    lower_url = url.lower()

    # Handle URLs without schemes
    candidate = url

    if candidate.startswith("//"):
        candidate = "http:" + candidate

    elif not re.match(
        r"^[a-zA-Z][a-zA-Z0-9+.-]*://",
        candidate
    ):
        candidate = "http://" + candidate

    parsed = urlsplit(candidate)

    # Extract hostname
    domain = (parsed.hostname or "").lower().rstrip(".")

    # Detect IP address
    try:
        ipaddress.ip_address(domain)
        has_ip = 1
    except ValueError:
        has_ip = 0

    # TLD extraction
    if has_ip or "." not in domain:
        tld = ""
    else:
        tld = domain.rsplit(".", 1)[-1]

    has_suspicious_tld = int(
        tld in SUSPICIOUS_TLDS
    )

    # Domain components
    domain_parts = domain.split(".") if domain else []

    num_subdomains = max(domain.count("."), 0)

    # Brand detection
    has_brand_keyword = int(
        any(brand in domain for brand in BRAND_KEYWORDS)
    )

    # Brand in leftmost subdomain
    brand_in_subdomain = int(
        len(domain_parts) > 2
        and any(
            brand in domain_parts[0]
            for brand in BRAND_KEYWORDS
        )
    )

    # Redirect indicators
    has_redirect = int(
        "redirect" in lower_url
        or "url=" in lower_url
    )

    return {
        "url_length": len(url),
        "domain_length": len(domain),
        "path_length": len(parsed.path),
        "query_length": len(parsed.query),
        "num_subdomains": num_subdomains,
        "has_ip_address": has_ip,
        "has_https": int(parsed.scheme.lower() == "https"),
        "has_suspicious_tld": has_suspicious_tld,
        "num_special_chars": sum(
            char in "-_@!%" for char in url
        ),
        "num_digits_in_domain": sum(
            char.isdigit() for char in domain
        ),
        "has_brand_keyword": has_brand_keyword,
        "brand_in_subdomain": brand_in_subdomain,
        "has_redirect": has_redirect,
        "path_depth": parsed.path.count("/")
    }


print("URL Feature Extraction Function Created!")
```

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Locate processed dataset
processed_path = Path(
    "../datasets/processed/url_features_14.csv"
)

if not processed_path.exists():
    raise FileNotFoundError(
        "Processed URL dataset not found. "
        "Complete feature extraction first."
    )

# Load processed dataset
processed_df = pd.read_csv(processed_path)

# Define exact feature order
URL_FEATURE_COLUMNS = [
    "url_length",
    "domain_length",
    "path_length",
    "query_length",
    "num_subdomains",
    "has_ip_address",
    "has_https",
    "has_suspicious_tld",
    "num_special_chars",
    "num_digits_in_domain",
    "has_brand_keyword",
    "brand_in_subdomain",
    "has_redirect",
    "path_depth"
]

# Prepare features and target
X_url = processed_df[URL_FEATURE_COLUMNS].astype(np.float32)

y_url = processed_df["target"].astype(np.int8)

# Validation
assert X_url.shape[1] == 14
assert X_url.isnull().sum().sum() == 0
assert np.isfinite(X_url.to_numpy()).all()
assert set(y_url.unique()) == {0, 1}

print("Dataset Loaded Successfully!")

print("Feature Shape:", X_url.shape)
print("Target Shape:", y_url.shape)

print("\nClass Distribution:")
print(y_url.value_counts())

print("\nFeature Validation: Passed")
```

```python
from sklearn.model_selection import train_test_split

X_train_url, X_test_url, y_train_url, y_test_url = train_test_split(
    X_url,
    y_url,
    test_size=0.20,
    random_state=42,
    stratify=y_url
)

print("Training Records:", len(X_train_url))
print("Testing Records:", len(X_test_url))

print("\nTraining Distribution:")
print(y_train_url.value_counts())

print("\nTesting Distribution:")
print(y_test_url.value_counts())
```

```python
from sklearn.ensemble import RandomForestClassifier
import time

# Initialize Random Forest
url_model = RandomForestClassifier(
    n_estimators=200,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1,
    verbose=1
)

print("URL Random Forest Training Started...")

start_time = time.time()

# Train model
url_model.fit(
    X_train_url,
    y_train_url
)

end_time = time.time()

print("\nTraining Completed Successfully!")

print("Number of Trees:", url_model.n_estimators)
print("Training Samples:", len(X_train_url))
print("Total Features:", url_model.n_features_in_)

print(
    f"Training Time: {(end_time-start_time)/60:.2f} minutes"
)
```

```python
# Generate predictions

y_pred_url = url_model.predict(X_test_url)

# Generate probabilities

y_prob_url = url_model.predict_proba(X_test_url)

print("Predictions Generated Successfully!")

print("Total Predictions:", len(y_pred_url))

print("\nFirst 10 Predictions:")
print(y_pred_url[:10])
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

accuracy_url = accuracy_score(y_test_url, y_pred_url)

precision_url = precision_score(
    y_test_url, y_pred_url, pos_label=1
)

recall_url = recall_score(
    y_test_url, y_pred_url, pos_label=1
)

f1_url = f1_score(
    y_test_url, y_pred_url, pos_label=1
)

print("===== URL MODEL EVALUATION =====")

print(f"Accuracy  : {accuracy_url:.4f}")
print(f"Precision : {precision_url:.4f}")
print(f"Recall    : {recall_url:.4f}")
print(f"F1-Score  : {f1_url:.4f}")

print("\nClassification Report:")

print(classification_report(
    y_test_url,
    y_pred_url,
    labels=[0, 1],
    target_names=["SAFE", "PHISHING"]
))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm_url = confusion_matrix(
    y_test_url,
    y_pred_url,
    labels=[0, 1]
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_url,
    display_labels=["SAFE", "PHISHING"]
)

disp.plot(cmap="Blues")

plt.title("URL Phishing Detection - Confusion Matrix")
plt.show()
```

```python
import joblib
import pandas as pd
import numpy as np
import json
import sklearn

from pathlib import Path

# File locations
model_path = Path("../Models/url_rf_model.pkl")
metadata_path = Path("../Models/url_feature_metadata.json")
dataset_path = Path("../datasets/processed/url_features_14.csv")

# Load model and metadata
url_model = joblib.load(model_path)

with open(metadata_path, "r") as file:
    metadata = json.load(file)

feature_columns = metadata["feature_columns"]

# Load processed dataset
df = pd.read_csv(dataset_path)

# Prepare actual URL features
X_sample = df[feature_columns].sample(
    n=min(10, len(df)),
    random_state=42
).astype(np.float32)

# Generate predictions
predictions = url_model.predict(X_sample)

probabilities = url_model.predict_proba(X_sample)

# Find positive class index
positive_index = list(url_model.classes_).index(1)

phishing_probabilities = probabilities[:, positive_index]

# Display results
print("===== URL MODEL COMPATIBILITY TEST =====")

print("Scikit-Learn Version:", sklearn.__version__)
print("Number of Trees:", url_model.n_estimators)
print("Total Features:", url_model.n_features_in_)
print("Feature Count:", len(feature_columns))

print("\nActual URL Predictions:")

for i, prediction in enumerate(predictions):

    label = "PHISHING" if prediction == 1 else "SAFE"

    print(
        f"URL {i+1}: {label} | "
        f"Phishing Probability: {phishing_probabilities[i]:.2%}"
    )

# Assertions
assert sklearn.__version__ == "1.9.1"
assert url_model.n_features_in_ == 14
assert X_sample.shape[1] == 14
assert probabilities.shape == (len(X_sample), 2)
assert np.isfinite(probabilities).all()

print("\nURL MODEL COMPATIBILITY: PASSED!")
```
